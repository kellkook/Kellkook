```python
import matplotlib.pyplot as plt
import numpy as np

# Dados das 8 vagas analisadas
habilidades = [
    "Linux",
    "Windows",
    "macOS",
    "Pacote Office",
    "Cloud Computing",
    "Automação / DevOps",
    "SOC",
    "Redes",
    "Firewall / WAF",
    "Frameworks de segurança",
    "Certificação AWS"
]

quantidade = [3, 2, 2, 2, 1, 1, 1, 1, 1, 1, 1]

# Criando uma paleta de cores
cores = [
    "#6C63FF",
    "#4ECDC4",
    "#45B7D1",
    "#96CEB4",
    "#FFD166",
    "#FF9F68",
    "#FF6B6B",
    "#A78BFA",
    "#F472B6",
    "#60A5FA",
    "#34D399"
]

# Criando o gráfico
fig, ax = plt.subplots(figsize=(11, 7))

barras = ax.barh(
    habilidades[::-1],
    quantidade[::-1],
    color=cores[::-1],
    height=0.65
)

# Adicionando os valores no final de cada barra
for barra in barras:
    valor = barra.get_width()

    ax.text(
        valor + 0.05,
        barra.get_y() + barra.get_height() / 2,
        str(int(valor)),
        va="center",
        fontsize=11,
        fontweight="bold"
    )

# Título e subtítulo
ax.set_title(
    "Hard Skills mais presentes nas vagas de TI",
    fontsize=19,
    fontweight="bold",
    pad=20
)

ax.text(
    0.5,
    1.01,
    "Análise de 8 vagas de Tecnologia da Informação",
    transform=ax.transAxes,
    ha="center",
    fontsize=11
)

# Nome dos eixos
ax.set_xlabel(
    "Quantidade de vagas",
    fontsize=12,
    labelpad=10
)

# Limites do eixo
ax.set_xlim(0, 3.6)

# Marcações do eixo X
ax.set_xticks([0, 1, 2, 3])

# Grade para facilitar a leitura
ax.grid(
    axis="x",
    linestyle="--",
    alpha=0.3
)

# Coloca a grade atrás das barras
ax.set_axisbelow(True)

# Remove as bordas desnecessárias
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["left"].set_visible(False)

# Ajustes visuais
ax.tick_params(axis="y", labelsize=10)
ax.tick_params(axis="x", labelsize=10)

# Rodapé
fig.text(
    0.5,
    0.01,
    "Fonte: levantamento realizado a partir das 8 vagas selecionadas.",
    ha="center",
    fontsize=9,
    color="#666666"
)

plt.tight_layout(rect=[0, 0.04, 1, 1])

# Mostra o gráfico
plt.show()
```
